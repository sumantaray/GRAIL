# GRAIL on 480 observed embryoid-body cells

Run in a kernel with the package installed. This is a short software demonstration on 96 cells from each collection window.

```python
from pathlib import Path
import sys, subprocess
import anndata as ad
import pandas as pd
root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
example = root / "datasets/example/eb_480.h5ad"
cells = ad.read_h5ad(example)
print(cells.shape)
print(cells.obs.groupby(["collection_window","t"], observed=True).size())
```

```python
subprocess.run([sys.executable, str(root / "scripts/directional_check.py")], check=True)
```

```python
from tempfile import mkdtemp
run = Path(mkdtemp(prefix="grail-example-"))
subprocess.run([sys.executable, "-m", "grail.cli", "run", "--data", str(example),
                "--config", str(root / "configs/quick.json"), "--out", str(run),
                "--seed", "42", "--save-observed"], check=True)
result = run / "lsh/seed-42"
pd.read_csv(result / "reconstruction.csv")
```

```python
display(pd.read_csv(result / "sensitivity.csv"))
profiles = ad.read_h5ad(result / "generated.h5ad")
print(profiles.shape, sorted(profiles.obs["t"].unique()))
```
